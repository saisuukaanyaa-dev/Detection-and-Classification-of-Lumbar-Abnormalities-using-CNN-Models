```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision.models import densenet121, densenet169, densenet201
import pandas as pd
import numpy as np
import pydicom
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt
import cv2
import os
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# Configuration
class Config:
    batch_size = 16
    learning_rate = 1e-4
    num_epochs = 100
    image_size = 384
    num_classes = 3  # normal_mild, moderate, severe
    densenet_version = '121'  # '121', '169', '201'

# Custom Dataset with improved DICOM handling
class SpineMRIDataset(Dataset):
    def __init__(self, df, image_dir, transform=None, mode='train'):
        self.df = df.reset_index(drop=True)
        self.image_dir = image_dir
        self.transform = transform
        self.mode = mode
        
        # Get all the target columns
        self.target_columns = [col for col in df.columns if any(condition in col for condition in 
                               ['spinal_canal_stenosis', 'neural_foraminal_narrowing', 'subarticular_stenosis'])]
    
    def __len__(self):
        return len(self.df)
    
    def __getitem__(self, idx):
        try:
            study_id = str(self.df.iloc[idx]['study_id'])
            series_id = str(self.df.iloc[idx]['series_id'])
            instance_number = str(int(self.df.iloc[idx]['instance_number']))
            
            # Load DICOM image
            image_path = os.path.join(self.image_dir, study_id, series_id, f"{instance_number}.dcm")
            
            if not os.path.exists(image_path):
                # Create a blank image if file doesn't exist
                image = np.zeros((512, 512), dtype=np.uint16)
            else:
                dicom = pydicom.dcmread(image_path)
                image = dicom.pixel_array
            
            # Handle different bit depths and normalize
            if image.dtype in [np.int16, np.uint16, np.int32]:
                # Normalize to 8-bit
                image = image.astype(np.float32)
                image = (image - image.min()) / (image.max() - image.min() + 1e-8)
                image = (image * 255).astype(np.uint8)
            
            # Convert to 3-channel if needed
            if len(image.shape) == 2:
                image = np.stack([image] * 3, axis=-1)
            elif len(image.shape) == 3 and image.shape[2] == 1:
                image = np.repeat(image, 3, axis=2)
            
            # Apply transforms
            if self.transform:
                image = self.transform(image)
            
            if self.mode == 'train':
                # Get all targets for this image
                targets = {}
                target_labels = {}  # For evaluation metrics
                for col in self.target_columns:
                    severity = self.df.iloc[idx][col]
                    # Handle NaN values
                    if pd.isna(severity):
                        targets[col] = torch.tensor([1.0, 0.0, 0.0], dtype=torch.float32)
                        target_labels[col] = 0  # normal_mild
                    elif severity == 'Normal/Mild':
                        targets[col] = torch.tensor([1.0, 0.0, 0.0], dtype=torch.float32)
                        target_labels[col] = 0
                    elif severity == 'Moderate':
                        targets[col] = torch.tensor([0.0, 1.0, 0.0], dtype=torch.float32)
                        target_labels[col] = 1
                    elif severity == 'Severe':
                        targets[col] = torch.tensor([0.0, 0.0, 1.0], dtype=torch.float32)
                        target_labels[col] = 2
                    else:
                        targets[col] = torch.tensor([1.0, 0.0, 0.0], dtype=torch.float32)
                        target_labels[col] = 0
                
                return image, targets, target_labels
            else:
                return image, study_id
                
        except Exception as e:
            # Return a dummy image and targets
            image = np.zeros((512, 512, 3), dtype=np.uint8)
            if self.transform:
                image = self.transform(image)
            
            if self.mode == 'train':
                targets = {}
                target_labels = {}
                for col in self.target_columns:
                    targets[col] = torch.tensor([1.0, 0.0, 0.0], dtype=torch.float32)
                    target_labels[col] = 0
                return image, targets, target_labels
            else:
                return image, "000000"

# Multi-Head DenseNet Model
class SpineDenseNet(nn.Module):
    def __init__(self, num_classes=3, version='121'):
        super(SpineDenseNet, self).__init__()
        
        # Backbone - using pretrained DenseNet
        if version == '121':
            self.backbone = densenet121(weights='DEFAULT')
            backbone_features = 1024
        elif version == '169':
            self.backbone = densenet169(weights='DEFAULT')
            backbone_features = 1664
        elif version == '201':
            self.backbone = densenet201(weights='DEFAULT')
            backbone_features = 1920
        else:
            raise ValueError("DenseNet version must be '121', '169', or '201'")
        
        # Remove the classifier
        self.backbone.classifier = nn.Identity()
        
        # Define all prediction heads for each condition and level
        self.heads = nn.ModuleDict()
        
        conditions = [
            'left_neural_foraminal_narrowing',
            'right_neural_foraminal_narrowing', 
            'left_subarticular_stenosis',
            'right_subarticular_stenosis',
            'spinal_canal_stenosis'
        ]
        
        levels = ['l1_l2', 'l2_l3', 'l3_l4', 'l4_l5', 'l5_s1']
        
        for condition in conditions:
            for level in levels:
                head_name = f"{condition}_{level}"
                self.heads[head_name] = nn.Sequential(
                    nn.Dropout(0.4),  # Slightly higher dropout for DenseNet
                    nn.Linear(backbone_features, 512),
                    nn.ReLU(inplace=True),
                    nn.BatchNorm1d(512),
                    nn.Dropout(0.3),
                    nn.Linear(512, 256),
                    nn.ReLU(inplace=True),
                    nn.BatchNorm1d(256),
                    nn.Dropout(0.2),
                    nn.Linear(256, num_classes)
                )
    
    def forward(self, x):
        features = self.backbone(x)
        
        outputs = {}
        for head_name, head in self.heads.items():
            outputs[head_name] = torch.softmax(head(features), dim=1)
            
        return outputs

# Weighted Loss Function
class WeightedMultiTaskLoss(nn.Module):
    def __init__(self):
        super(WeightedMultiTaskLoss, self).__init__()
        
    def forward(self, outputs, targets):
        total_loss = 0
        num_valid_targets = 0
        
        for key in outputs.keys():
            if key in targets:
                # Competition weights: normal_mild=1, moderate=2, severe=4
                competition_weights = torch.tensor([1.0, 2.0, 4.0]).to(outputs[key].device)
                
                # Use targets directly (they're already one-hot encoded)
                target_class = torch.argmax(targets[key], dim=1)
                loss = nn.CrossEntropyLoss(weight=competition_weights)(outputs[key], target_class)
                total_loss += loss
                num_valid_targets += 1
        
        return total_loss / max(num_valid_targets, 1)

# Function to calculate metrics
def calculate_metrics(all_predictions, all_targets):
    """Calculate accuracy, confusion matrix, and classification report"""
    # Flatten all predictions and targets
    all_preds_flat = []
    all_targets_flat = []
    
    for condition_level in all_predictions.keys():
        if condition_level in all_targets and len(all_predictions[condition_level]) > 0:
            all_preds_flat.extend(all_predictions[condition_level])
            all_targets_flat.extend(all_targets[condition_level])
    
    if len(all_preds_flat) == 0:
        print("No predictions to evaluate!")
        return 0.0, None, None
    
    # Calculate accuracy
    accuracy = accuracy_score(all_targets_flat, all_preds_flat)
    
    # Calculate confusion matrix
    cm = confusion_matrix(all_targets_flat, all_preds_flat, labels=[0, 1, 2])
    
    # Calculate classification report
    class_names = ['Normal/Mild', 'Moderate', 'Severe']
    report = classification_report(all_targets_flat, all_preds_flat, 
                                  target_names=class_names, output_dict=True)
    
    return accuracy, cm, report

# Function to plot confusion matrix
def plot_confusion_matrix(cm, accuracy, epoch):
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Normal/Mild', 'Moderate', 'Severe'],
                yticklabels=['Normal/Mild', 'Moderate', 'Severe'])
    plt.title(f'Confusion Matrix - Epoch {epoch}\nAccuracy: {accuracy:.4f}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.tight_layout()
    plt.savefig(f'confusion_matrix_epoch_{epoch}.png')
    plt.show()

# Function to print detailed metrics
def print_detailed_metrics(report, accuracy):
    print("\n" + "="*60)
    print("DETAILED CLASSIFICATION METRICS")
    print("="*60)
    print(f"Overall Accuracy: {accuracy:.4f}")
    print("\nPer-class Metrics:")
    print("-" * 40)
    print(f"{'Class':<12} {'Precision':<10} {'Recall':<10} {'F1-Score':<10} {'Support':<10}")
    print("-" * 40)
    
    for class_name in ['Normal/Mild', 'Moderate', 'Severe']:
        if class_name in report:
            metrics = report[class_name]
            print(f"{class_name:<12} {metrics['precision']:.4f}     {metrics['recall']:.4f}     "
                  f"{metrics['f1-score']:.4f}     {metrics['support']:<10}")
    
    # Print macro and weighted averages
    if 'macro avg' in report:
        macro = report['macro avg']
        weighted = report['weighted avg']
        print("-" * 40)
        print(f"{'Macro Avg':<12} {macro['precision']:.4f}     {macro['recall']:.4f}     "
              f"{macro['f1-score']:.4f}     {macro['support']:<10}")
        print(f"{'Weighted Avg':<12} {weighted['precision']:.4f}     {weighted['recall']:.4f}     "
              f"{weighted['f1-score']:.4f}     {weighted['support']:<10}")
    print("="*60)

# Training function with metrics
def train_model(model, train_loader, val_loader, criterion, optimizer, scheduler, num_epochs, device):
    best_val_loss = float('inf')
    best_val_accuracy = 0.0
    
    # Lists to store metrics for plotting
    train_losses = []
    val_losses = []
    train_accuracies = []
    val_accuracies = []
    
    for epoch in range(num_epochs):
        # Training phase
        model.train()
        train_loss = 0
        train_bar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs} [Train]')
        
        # For storing training metrics
        train_predictions = {}
        train_targets = {}
        
        for batch_idx, (images, targets, target_labels) in enumerate(train_bar):
            try:
                images = images.to(device)
                
                # Move targets to device
                batch_targets = {}
                for key, value in targets.items():
                    batch_targets[key] = value.to(device)
                
                optimizer.zero_grad()
                outputs = model(images)
                loss = criterion(outputs, batch_targets)
                loss.backward()
                optimizer.step()
                
                train_loss += loss.item()
                train_bar.set_postfix(loss=loss.item())
                
                # Store predictions and targets for metrics
                with torch.no_grad():
                    for key in outputs.keys():
                        if key in target_labels:
                            preds = torch.argmax(outputs[key], dim=1).cpu().numpy()
                            if key not in train_predictions:
                                train_predictions[key] = []
                                train_targets[key] = []
                            
                            # Get batch target labels
                            batch_target_labels = []
                            for i in range(len(images)):
                                if key in target_labels:
                                    batch_target_labels.append(target_labels[key][i])
                            
                            train_predictions[key].extend(preds)
                            train_targets[key].extend(batch_target_labels)
                
            except Exception as e:
                continue
        
        # Calculate training metrics
        train_accuracy, train_cm, train_report = calculate_metrics(train_predictions, train_targets)
        
        # Validation phase
        model.eval()
        val_loss = 0
        val_bar = tqdm(val_loader, desc=f'Epoch {epoch+1}/{num_epochs} [Val]')
        
        # For storing validation metrics
        val_predictions = {}
        val_targets = {}
        
        with torch.no_grad():
            for batch_idx, (images, targets, target_labels) in enumerate(val_bar):
                try:
                    images = images.to(device)
                    
                    batch_targets = {}
                    for key, value in targets.items():
                        batch_targets[key] = value.to(device)
                    
                    outputs = model(images)
                    loss = criterion(outputs, batch_targets)
                    val_loss += loss.item()
                    val_bar.set_postfix(loss=loss.item())
                    
                    # Store predictions and targets for metrics
                    for key in outputs.keys():
                        if key in target_labels:
                            preds = torch.argmax(outputs[key], dim=1).cpu().numpy()
                            if key not in val_predictions:
                                val_predictions[key] = []
                                val_targets[key] = []
                            
                            # Get batch target labels
                            batch_target_labels = []
                            for i in range(len(images)):
                                if key in target_labels:
                                    batch_target_labels.append(target_labels[key][i])
                            
                            val_predictions[key].extend(preds)
                            val_targets[key].extend(batch_target_labels)
                    
                except Exception as e:
                    continue
        
        # Calculate validation metrics
        val_accuracy, val_cm, val_report = calculate_metrics(val_predictions, val_targets)
        
        avg_train_loss = train_loss / len(train_loader)
        avg_val_loss = val_loss / len(val_loader)
        
        # Store metrics for plotting
        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)
        train_accuracies.append(train_accuracy)
        val_accuracies.append(val_accuracy)
        
        print(f'\nEpoch {epoch+1}/{num_epochs}:')
        print(f'Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}')
        print(f'Train Accuracy: {train_accuracy:.4f}, Val Accuracy: {val_accuracy:.4f}')
        
        # Print detailed metrics
        if val_report:
            print_detailed_metrics(val_report, val_accuracy)
        
        # Plot confusion matrix
        if val_cm is not None:
            plot_confusion_matrix(val_cm, val_accuracy, epoch+1)
        
        # Learning rate scheduling
        scheduler.step(avg_val_loss)
        
        # Save best model
        if avg_val_loss < best_val_loss or val_accuracy > best_val_accuracy:
            best_val_loss = min(avg_val_loss, best_val_loss)
            best_val_accuracy = max(val_accuracy, best_val_accuracy)
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'val_loss': avg_val_loss,
                'val_accuracy': val_accuracy,
            }, 'best_spine_densenet.pth')
            print(f'💾 Best model saved! Val Loss: {avg_val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}')
        
        print("-" * 80)
    
    # Plot training history
    plot_training_history(train_losses, val_losses, train_accuracies, val_accuracies)

# Function to plot training history
def plot_training_history(train_losses, val_losses, train_accuracies, val_accuracies):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Plot losses
    ax1.plot(train_losses, label='Train Loss', marker='o')
    ax1.plot(val_losses, label='Val Loss', marker='o')
    ax1.set_title('Training and Validation Loss')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.legend()
    ax1.grid(True)
    
    # Plot accuracies
    ax2.plot(train_accuracies, label='Train Accuracy', marker='o')
    ax2.plot(val_accuracies, label='Val Accuracy', marker='o')
    ax2.set_title('Training and Validation Accuracy')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Accuracy')
    ax2.legend()
    ax2.grid(True)
    
    plt.tight_layout()
    plt.savefig('training_history.png')
    plt.show()

# Data transforms
def get_transforms():
    train_transform = transforms.Compose([
        transforms.ToPILImage(),
        transforms.Resize((Config.image_size, Config.image_size)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(10),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    val_transform = transforms.Compose([
        transforms.ToPILImage(),
        transforms.Resize((Config.image_size, Config.image_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    return train_transform, val_transform

# Main training pipeline
def main():
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f'Using device: {device}')
    
    try:
        # Load data
        train_df = pd.read_csv('/kaggle/input/rsna-2024-lumbar-spine-degenerative-classification/train.csv')
        train_coords = pd.read_csv('/kaggle/input/rsna-2024-lumbar-spine-degenerative-classification/train_label_coordinates.csv')
        
        print(f"Train CSV shape: {train_df.shape}")
        print(f"Train coordinates shape: {train_coords.shape}")
        
        # Merge data
        df = pd.merge(train_coords, train_df, on='study_id', how='left')
        print(f"Merged data shape: {df.shape}")
        
        # Check class distribution
        target_columns = [col for col in df.columns if any(condition in col for condition in 
                          ['spinal_canal_stenosis', 'neural_foraminal_narrowing', 'subarticular_stenosis'])]
        
        print(f"\nNumber of target columns: {len(target_columns)}")
        
        # Sample a smaller subset for testing
        if len(df) > 1000:
            df = df.sample(1000, random_state=42)
            print(f"Using subset of {len(df)} samples for testing")
        
        # Split data
        train_data, val_data = train_test_split(df, test_size=0.2, random_state=42)
        
        print(f"Train data: {len(train_data)}, Val data: {len(val_data)}")
        
        # Get transforms
        train_transform, val_transform = get_transforms()
        
        # Create datasets
        train_dataset = SpineMRIDataset(train_data, 
                                      '/kaggle/input/rsna-2024-lumbar-spine-degenerative-classification/train_images', 
                                      train_transform, 'train')
        val_dataset = SpineMRIDataset(val_data, 
                                    '/kaggle/input/rsna-2024-lumbar-spine-degenerative-classification/train_images', 
                                    val_transform, 'train')
        
        # Create data loaders
        train_loader = DataLoader(train_dataset, batch_size=Config.batch_size, shuffle=True, num_workers=2)
        val_loader = DataLoader(val_dataset, batch_size=Config.batch_size, shuffle=False, num_workers=2)
        
        # Initialize model
        model = SpineDenseNet(num_classes=Config.num_classes, version=Config.densenet_version).to(device)
        print(f"✅ DenseNet-{Config.densenet_version} Model initialized")
        
        # Print model summary
        total_params = sum(p.numel() for p in model.parameters())
        trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
        print(f"📊 Total parameters: {total_params:,}")
        print(f"📊 Trainable parameters: {trainable_params:,}")
        
        # Loss and optimizer
        criterion = WeightedMultiTaskLoss()
        optimizer = optim.AdamW(model.parameters(), lr=Config.learning_rate, weight_decay=1e-4)
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=3, factor=0.5, verbose=True)
        
        print("🚀 Starting training...")
        # Train model
        train_model(model, train_loader, val_loader, criterion, optimizer, scheduler, Config.num_epochs, device)
        
    except Exception as e:
        print(f"❌ Error in main: {e}")
        import traceback
        traceback.print_exc()

if __name__ == '__main__':
    main()
```
